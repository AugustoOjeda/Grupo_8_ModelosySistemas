El presente informe tiene como objetivo estudiar el comportamiento de los contaminantes incorporados a los sistemas hidrológicos como consecuencia de la actividad humana, ya sea de manera intencional o accidental. En particular, se busca comprender los mecanismos mediante los cuales estas sustancias son transportadas, transformadas y acumuladas en el medio acuático, con el propósito de desarrollar un modelo matemático que permita analizar y predecir su evolución temporal y sus posibles efectos sobre la calidad del agua.

Para abordar esta problemática, se toma como referencia el trabajo desarrollado por H. B. Fischer, E. J. List, R. C. Y. Koh, J. Imberger y N. H. Brooks en Mixing in Inland and Coastal Waters (1979). A partir de los principios de conservación de la materia aplicados al transporte de contaminantes, es posible formular un modelo simplificado mediante la siguiente expresión general:

$$
\text{Acumulación} = \text{Entrada} - \text{Salida} + \text{Generación} - \text{Eliminación}
\tag{1}
$$

Esta relación establece que la variación de la cantidad de contaminante presente en un sistema depende del balance entre los aportes que recibe, las pérdidas asociadas al transporte y los procesos internos de generación o eliminación.

Para obtener una representación matemática del sistema, se considera un reactor perfectamente mezclado, de volumen constante y con procesos de eliminación que pueden describirse mediante una cinética de primer orden. Bajo estas hipótesis, la concentración del contaminante es uniforme en todo el volumen del reactor, por lo que la masa total presente puede expresarse como $M=VC$.

Aplicando el balance de materia de la ecuación (1), se obtiene:

$$
\frac{d(VC)}{dt}=QC_{\text{in}}-QC+VS-kVC
\tag{2}
$$

Si se supone que el fluido entrante se encuentra libre de contaminantes, es decir, $C_{\text{in}}=0$, y que el volumen permanece constante, la ecuación (2) puede simplificarse para obtener la siguiente ecuación diferencial:

$$
\dot{C}=-\frac{Q}{V}C-kC+S
\tag{3}
$$

En esta expresión, el término $\frac{Q}{V}C$ representa la disminución de la concentración debido al transporte del contaminante hacia el exterior del reactor. Por su parte, $kC$ describe la reducción de la concentración asociada a los procesos de degradación o remoción, mientras que $S$ representa el incremento producido por una fuente contaminante interna.

De esta manera, la ecuación (3) permite describir la evolución temporal de la concentración a partir de la interacción entre los mecanismos de transporte, eliminación y generación del contaminante.

Como formulación complementaria, el balance general puede expresarse directamente en términos de la masa total del contaminante presente en el sistema.

Considerando inicialmente el caso particular en el que no existen fuentes internas de generación, la ecuación (1) se reduce a:

$$
\frac{dM}{dt}=\text{Flujo másico entrante}-\text{Flujo másico saliente}-\text{Tasa de eliminación}
\tag{4}
$$

Para desarrollar esta expresión, se considera que el fluido ingresa con una concentración $C_{\text{in}}$, mientras que el flujo de salida presenta la misma concentración $C$ que el interior del reactor. Asimismo, se supone que ambos flujos poseen un caudal volumétrico $Q$ y que la eliminación del contaminante responde a una cinética de primer orden, caracterizada por una constante $k_R$.

Bajo estas condiciones, el balance de masa queda expresado como:

$$
\dot{M}=Q(C_{\text{in}}-C)-k_RM
\tag{5}
$$

Esta ecuación permite analizar directamente la variación temporal de la masa del contaminante, considerando los intercambios con el entorno y los procesos de eliminación que tienen lugar dentro del sistema.

Finalmente, resulta necesario incorporar la evolución temporal de la fuente contaminante. Para ello, se considera que su intensidad puede disminuir como consecuencia de procesos de agotamiento o decaimiento, los cuales se representan mediante una ley de primer orden.

La hipótesis matemática que describe este comportamiento es:

$$
\frac{dS}{dt}=-\lambda S
\tag{6}
$$

Donde $\lambda$ representa una constante positiva de decaimiento y $S$ corresponde a la intensidad de la fuente contaminante.

Esta expresión establece que la velocidad de disminución de la fuente es proporcional a su intensidad instantánea. En consecuencia, a medida que disminuye el valor de $S$, también se reduce la magnitud de su tasa de decaimiento.

Sin embargo, este comportamiento únicamente resulta válido cuando no existen mecanismos externos que modifiquen la intensidad de la fuente. Para contemplar la posibilidad de nuevos aportes o perturbaciones externas, se incorpora una función temporal $u(t)$, que representa la acción externa sobre dicha fuente.

De esta manera, la ecuación diferencial que describe su comportamiento adopta la forma:

$$
\dot{S}=-\lambda S+u(t)
\tag{7}
$$

La incorporación de $u(t)$ permite representar diferentes escenarios de contaminación, incluyendo situaciones en las que la fuente recibe aportes adicionales o experimenta variaciones a lo largo del tiempo.
