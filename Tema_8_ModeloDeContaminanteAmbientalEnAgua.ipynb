**Tema:** Modelo de Contaminante Ambiental en Agua (Adaptación No Lineal)

**Nota sobre la adaptación del modelo:** 
Para cumplir estrictamente con los requisitos de la cátedra respecto al análisis de sistemas no lineales, se modificó el modelo lineal original de degradación genérica. Se adoptó un modelo de **cinética de saturación biológica de Michaelis-Menten**, asumiendo a degradación de materia orgánica (medida como DBO - Demanda Biológica de Oxígeno) mediante bacterias (lodos activados). Los parámetros biológicos extraídos de la literatura ($k_{max}$ y $K_m$) fueron seleccionados para que su linealización a bajas concentraciones respete la constante de decaimiento original del sistema ($k = 0.05 \text{ min}^{-1}$). Las dimensiones y caudales del reactor original se mantuvieron intactos.

## 1. Modelo de Contaminante en Reactor (No Lineal)

Se considera un reactor perfectamente mezclado en el que un contaminante orgánico es transportado, eliminado biológicamente y generado mediante una fuente dinámica:

$$ \dot{C} = -\frac{Q}{V}C - \frac{k_{max} C}{K_m + C} + S $$
$$ \dot{M} = Q(C_{in} - C) - k_r M $$
$$ \dot{S} = -\lambda S + u(t) $$

**Donde:**
*   $C$: Concentración del contaminante (DBO).
*   $M$: Masa acumulada en el sistema.
*   $S$: Fuente contaminante variable.

### 1.1 Valores de Parámetros y Condiciones Iniciales

**Condiciones Iniciales:**
$C(0) = 0.5 \text{ mg L}^{-1}$  |  $M(0) = 1 \text{ mg}$  |  $S(0) = 0.5 \text{ mg L}^{-1}\text{min}^{-1}$

**Tabla de Parámetros:**

| Parámetro | Valor | Unidad | Descripción |
| :--- | :--- | :--- | :--- |
| $Q$ | 1 | $\text{L min}^{-1}$ | Caudal de entrada/salida |
| $V$ | 10 | $\text{L}$ | Volumen del reactor |
| $C_{in}$ | 1 | $\text{mg L}^{-1}$ | Concentración de entrada |
| $k_{max}$ | 1.0 | $\text{mg L}^{-1}\text{min}^{-1}$ | Tasa máxima de asimilación bacteriana |
| $K_m$ | 20 | $\text{mg L}^{-1}$ | Constante de semisaturación biológica |
| $k_r$ | 0.02 | $\text{min}^{-1}$ | Constante de remoción de masa |
| $\lambda$ | 0.1 | $\text{min}^{-1}$ | Tasa de decaimiento de la fuente |

**Para la simulación temporal:**
*   $t \in [0, 100] \text{ min}$
*   $N = 5000$ muestras
*   Entrada constante: $u(t) = 0.2 \text{ mg min}^{-2}$
